#📊 Carga y análisis

```python
import numpy as np
dt1 = data['Churn'].str.strip()
dt1 = dt1.replace(r'^\s*$', np.nan, regex=True)
dt1.value_counts()
```

```python
import matplotlib.pyplot as plt

conteo_categorias = dt1.value_counts()
categorias = conteo_categorias.index
frecuencias = conteo_categorias.values

plt.figure(figsize=(4, 2))
plt.bar(categorias, frecuencias, color='green')

plt.title('Frecuencia')
plt.xlabel('Categoría')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig1.png')
plt.tight_layout()
plt.show()
```

```python
dt2=pd.json_normalize(data['customer'])
dtgender=dt2['gender']
dtchurn= data['Churn']
dtgender1=dtgender.tolist()
dtchurn1=dtchurn.tolist()
dtnuevo=pd.DataFrame({'gender':dtgender1,'Churn':dtchurn1})
dtnuevo = dtnuevo.replace(r'^\s*$', np.nan, regex=True)
dtnuevo
```

```python
import seaborn as sns

df = pd.DataFrame(dtnuevo)

print("DataFrame original:")
print(df)
print("-" * 30)

conteo_cruzado = pd.crosstab(df['Churn'], df['gender'])

print("\nTabla de contingencia (conteo cruzado):")
print(conteo_cruzado)
print("-" * 30)
```

```python
plt.figure(figsize=(8, 6))
figura2=plt.figure(figsize=(8, 6))

conteo_cruzado.plot(kind='bar', ax=plt.gca(), rot=0, color=['skyblue', 'lightcoral'])

plt.title('Distribución de Género por Cancelación', fontsize=14)
plt.xlabel('Cancelación', fontsize=12)
plt.ylabel('Conteo', fontsize=12)
plt.xticks(fontsize=10) # 'yes' y 'no'
plt.yticks(fontsize=10)
plt.legend(title='Género', fontsize=10, title_fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig2.png')
plt.tight_layout()
plt.show()
```

```python
dt3=pd.json_normalize(data['account'])
dtcontrato=dt3['Contract']
dtchurn= data['Churn']
dtcontrato1=dtcontrato.tolist()
dtchurn1=dtchurn.tolist()
dtnuevo1=pd.DataFrame({'Contract':dtcontrato1,'Churn':dtchurn1})
dtnuevo1 = dtnuevo1.replace(r'^\s*$', np.nan, regex=True)
dtnuevo1
```

```python
df1 = pd.DataFrame(dtnuevo1)

print("DataFrame original:")
print(df1)
print("-" * 30)

conteo_cruzado1 = pd.crosstab(df1['Churn'], df1['Contract'])

print("\nTabla de contingencia (conteo cruzado):")
print(conteo_cruzado1)
print("-" * 30)
```

```python
plt.figure(figsize=(8, 6))
conteo_cruzado1.plot(kind='bar', ax=plt.gca(), rot=0, color=['Plum', 'Blueviolet', 'Blue'])

plt.title('Distribución de Contrato por Cancelación', fontsize=14)
plt.xlabel('Cancelación', fontsize=12)
plt.ylabel('Conteo', fontsize=12)
plt.xticks(fontsize=10) # 'yes' y 'no'
plt.yticks(fontsize=10)
plt.legend(title='Contrato', fontsize=10, title_fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig3.png')
plt.tight_layout()
plt.show()
```

```python
dt4=pd.json_normalize(data['account'])
dtpago=dt4['PaymentMethod']
dtchurn= data['Churn']
dtpago=dtpago.tolist()
dtchurn1=dtchurn.tolist()
dtnuevo2=pd.DataFrame({'PaymentMethod':dtpago,'Churn':dtchurn1})
dtnuevo2 = dtnuevo2.replace(r'^\s*$', np.nan, regex=True)
dtnuevo2
```

```python
df2 = pd.DataFrame(dtnuevo2)

print("DataFrame original:")
print(df2)
print("-" * 30)

conteo_cruzado2 = pd.crosstab(df2['Churn'], df2['PaymentMethod'])

print("\nTabla de contingencia (conteo cruzado):")
print(conteo_cruzado2)
print("-" * 30)
```

```python
plt.figure(figsize=(8, 6))
conteo_cruzado2.plot(kind='bar', ax=plt.gca(), rot=0, color=['Plum', 'Blueviolet', 'Blue', 'Brown', 'Black'])

plt.title('Distribución de Método de Pago por Cancelación', fontsize=14)
plt.xlabel('Cancelación', fontsize=12)
plt.ylabel('Conteo', fontsize=12)
plt.xticks(fontsize=10) # 'yes' y 'no'
plt.yticks(fontsize=10)
plt.legend(title='Contrato', fontsize=10, title_fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig4.png')
plt.tight_layout()
plt.show()
```

```python
dt5=pd.json_normalize(data['account'])
dtmensual=dt5['Charges.Monthly']
dtchurn= data['Churn']
dtmensual=dtmensual.tolist()
dtchurn1=dtchurn.tolist()
dtnuevo3=pd.DataFrame({'Charges.monthly':dtmensual,'Churn':dtchurn1})
dtnuevo3 = dtnuevo3.replace(r'^\s*$', np.nan, regex=True)
dtnuevo3.info()
```

```python
plt.figure(figsize=(9, 6))

sns.boxplot(x=dtchurn, y=dtmensual, data=dtnuevo3, palette='Set2', hue=dtchurn)

plt.title('Cancelación por cargos mensuales', fontsize=16)
plt.xlabel('Cancelación', fontsize=12)
plt.ylabel('Cargos Mensuales', fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig5.png')
plt.tight_layout()
plt.show()
```

```python
dt6=pd.json_normalize(data['customer'])
dtenure=dt6['tenure']
plt.figure(figsize=(9, 6))

sns.boxplot(x=dtchurn, y=dtenure, data=dt, palette='Set2', hue=dtchurn)

plt.title('Cancelación por ciclo de contrato', fontsize=16)
plt.xlabel('Cancelación', fontsize=12)
plt.ylabel('Ciclo de Contrato', fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig('fig6.png')
plt.tight_layout()
plt.show()
```
